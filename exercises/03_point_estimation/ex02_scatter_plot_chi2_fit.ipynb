# Scatter plot ($\chi^2$ fit)

Lecture 4 (25/11) - Point estimation

## Exercise

1. Generate $N = 11$ points with coordinates

   $$x_i = 0, 100, 200, \dots, 1000, \qquad y_i = m\,x_i + q + \delta y_i$$

   where $m = 1.5$, $q = 3$, $\delta y_i$ is a random Gaussian number with $\mu = 0$ and standard deviation $\sigma_i$, and $\sigma_i$ is a random number uniformly distributed in $[10, 100[$.

   To each point assign the uncertainties $\sigma_{x_i} = 0$ and $\sigma_{y_i} = \sigma_i$.
2. Fit the scatter plot with the $\chi^2$ fit, and retrieve the minimum $\chi^2$ from the fitter. (In ROOT the fit result gives it through `Chi2()`.)
3. Repeat for $10^4$ toy-MC data, and plot the histogram of the minimum $\chi^2$.
4. Overlay with the theoretical $\chi^2$ distribution with the same number of degrees of freedom: $n_{\text{DOF}} = n - 2$ (because we have 2 parameters).
