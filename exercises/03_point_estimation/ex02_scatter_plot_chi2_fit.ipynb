# Scatter plot ($\chi^2$ fit)

Lecture 5 (26/11) - Point estimation

## Exercise

1. Generate $N = 11$ points with the following coordinates:

   $$x_i = 0, 100, 200, \dots, 1000, \qquad y_i = m\,x_i + q + \delta y_i$$

   where $m = 1.5$, $q = 3$, $\delta y_i$ is a random Gaussian ($\mu = 0$, $\sigma_i$) and $\sigma_i$ is a random uniform in $[10, 100[$.

   To each point, assign the uncertainties: $\sigma_{x_i} = 0$, $\sigma_{y_i} = \sigma_i$.
2. Fit the scatter plot with the $\chi^2$ fit, and retrieve the minimum $\chi^2$ from the fitter.

   In ROOT:

   ```cpp
   TFitResultPtr resptr = scatterplot->Fit(...);
   TFitResult* res = resptr.Get();
   double chi2 = res->Chi2();
   ```
3. Repeat for $10^4$ toy-MC data, and plot the histogram of min-$\chi^2$.
4. Overlay with the theoretical $\chi^2$ distribution with the same number of DOF: $n_{\text{DOF}} = n - 2$ (because we have 2 parameters).
