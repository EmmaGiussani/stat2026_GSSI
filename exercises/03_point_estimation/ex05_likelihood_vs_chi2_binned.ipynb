# Likelihood vs $\chi^2$ for binned Poisson data

Lecture 5 (26/11) - Point estimation

## Exercise

1. Generate data distributed with the following signal and background PDFs.

   The energy $E$ is in keV:
   
   $$f_s(E) = \frac{1}{\sqrt{2\pi}\,\sigma} \exp\!\left(-\frac{(E-\mu)^2}{2\sigma^2}\right), \qquad \mu = 2039\ \text{keV},\quad \sigma = 1.5\ \text{keV}$$
   
   $$f_b(E) = \frac{1}{E_{\max} - E_{\min}}, \qquad E_{\min} = 2000\ \text{keV},\quad E_{\max} = 2080\ \text{keV}$$
   
   where $[E_{\min}, E_{\max}]$ is the fit range.

   Generate 300 toy-MC data using all combinations of the following number of signal and background counts:
   - $s = \{10,\ 100,\ 1000,\ 10\,000\}$
   - $b = \{10,\ 100,\ 1000,\ 10\,000\}$

   Bin the data with 0.1 keV binning.
2. Fit each histogram with $\mathcal{L}$ (Poisson likelihood), $\chi^2_N$ (Neyman) and $\chi^2_P$ (Pearson).
3. Produce the following plots:
   1. Distribution of the best fit of $s$ for all combinations of $s$ and $b$, with the distributions obtained from the 3 methods overlapped together.
   2. Same, but for $b$.
   3. 2-dim distribution of $\hat{s}$ vs $\hat{b}$ for all combinations of $s$ and $b$. This time, do 3 separate plots for $\mathcal{L}$, $\chi^2_N$, $\chi^2_P$, otherwise it's a mess.
