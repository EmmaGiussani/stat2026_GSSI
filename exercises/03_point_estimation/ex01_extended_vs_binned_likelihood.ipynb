# Extended vs binned likelihood

Lecture 5 (26/11) - Point estimation

## Exercise

1. Generate data distributed with the following signal and background PDFs:

   $$f_s(E) = \frac{1}{\sqrt{2\pi}\,\sigma} \exp\!\left(-\frac{(E-\mu)^2}{2\sigma^2}\right), \qquad \mu = 2039\ \text{keV},\quad \sigma = 1.5\ \text{keV}$$

   $$f_b(E) = \frac{1}{E_{\max} - E_{\min}}, \qquad E_{\min} = 2000\ \text{keV},\quad E_{\max} = 2080\ \text{keV}$$

   (this is our fit range). Generate $10^4$ toy-MC data using all combinations of the following number of signal and background counts:
   - $S = \{5, 20, 50, 100\}$
   - $B = \{10, 30, 100, 200\}$
2. Fit each toy-MC data with the extended likelihood (Poisson).
3. Bin the toy-MC data with 1 keV binning, and fit the spectra with the binned likelihood (Poisson term for each bin).
4. Extract the CPU time required for each fit.
5. Plot the average of $\dfrac{t_{\text{extended}}}{t_{\text{binned}}}$ vs $\dfrac{N_{\text{events}}}{N_{\text{bins}}}$.

*Hint from the lecture:* for example, the number of signal events is a Poisson number with $s = 5$, and the events are then distributed with $f_s$; the number of background events is a Poisson number with $b = 10$, and the events are then distributed with $f_B$.
