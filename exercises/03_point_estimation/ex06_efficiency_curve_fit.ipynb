# Efficiency curve fit with binomial $\mathcal{L}$, Poisson $\mathcal{L}$ and Gaussian $\chi^2$

Lecture 4 (25/11) - Point estimation

## Exercise

1. Generate 20 points with coordinates $(E_i, k_i)$:

   $$E_i = 1, 2, 3, \dots, 20\ \text{keV}, \qquad \varepsilon_i = \frac{\Sigma}{2}\left[1 + \operatorname{erf}\!\left(\frac{E_i - \mu}{\sigma}\right)\right]$$

   with $\Sigma = 0.9,\ 0.95,\ 0.99$, $\mu = 3$ keV and $\sigma = 1$ keV.

   For each $\varepsilon_i$, generate $n$ random values uniformly distributed in $[0, 1[$ and set $k_i$ to the number of values $< \varepsilon_i$. Consider the following values for $n$: $100,\ 1000,\ 10\,000$.
2. Fit the data assuming the following distributions for $k_i$:
   1. Binomial
   2. Poisson
   3. $\chi^2$ (Neyman or Pearson, your choice)
3. Repeat for $10^3$ toys for all combinations of $n$ and $\Sigma$.
4. Plot the best-fit distribution of $\hat{\Sigma}$ obtained with the 3 methods, for all combinations of $\Sigma$ and $n$.

This example is the case of a pulse generator injecting $n$ artificial pulses into a detector, with 20 different energy values between 1 and 20 keV. $\Sigma$ is the overall trigger efficiency, $\mu$ is the trigger threshold, and $\sigma$ describes the rise of the efficiency curve.

**Note:** on the original sheet the symbol of the plateau efficiency is overwritten; here it is called $\Sigma$, as in the sketch of the efficiency curve.
