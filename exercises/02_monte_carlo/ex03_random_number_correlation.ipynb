# Random number correlation

Lecture 3 (17/11) - Monte Carlo methods

## Exercise

1. Generate $(10^5 + 1)$ random numbers distributed as a Gaussian with $\mu = 3$ and $\sigma = 2$.
2. Plot the distribution of $x_i$ and of $(x_i - x_{i-1})$.
3. Do the same, but this time generate the numbers $x_i$ using Metropolis-Hastings, until you accept $(10^5 + 1)$ points. As a proposal function, use a Gaussian with $\mu = x_i$ and $\sigma = 0.3$.

How do the distributions look like? Can you explain the standard deviation of the $(x_i - x_{i-1})$ distributions in the two cases?
