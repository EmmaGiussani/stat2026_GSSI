# Neyman belt for Binomial

Lecture 6 - Interval estimation

## Exercise

Assume we have a detector injected with $n$ pulser events, and that we want to measure the trigger efficiency $p$.

Take:
- $n$ = # of injected events = $\{0, 1, 2, \dots, 10\}$
- $p$ = input trigger probability = $0,\ 10^{-3},\ 2\cdot 10^{-3},\ \dots,\ 1$

1. Create a 2-dim histogram for $(k, p)$. Notice that $0 \le k \le n$, so $k$ must go from 0 to 10.
2. For each value of $p$, populate the corresponding distribution of $k$ with a Binomial$(k\,|\,n, p)$.
3. For each value of $p$, extract the shortest 68% interval.
4. Plot the Neyman confidence belt.
5. Compute the coverage of $p$.
