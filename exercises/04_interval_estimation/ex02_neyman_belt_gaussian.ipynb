1. Create a 2-dim histogram of $(x, \theta)$ with $0 \le x \le 10$ and $0 \le \theta \le 20$.
2. For each value of $\theta$, populate the corresponding $x$ with $10^5$ values of $x$ randomly sampled from $f(x|\theta)$.
3. Invert the belt for $x = 5$, and plot the corresponding $f(\theta)$: find $[\theta_1, \theta_2]$.
4. Fit $f(\theta)$ with a Gaussian.
5. Plot the coverage.
