## Exercise

Suppose to have a variable $\lambda$ with PDF $f(\lambda) = \text{Poisson}(n, \lambda) = \dfrac{e^{-\lambda}\lambda^n}{n!}$ with $n = 3$.

Notice that we are looking at the Poisson as a function of the expectation value, not of the number of measured counts!

1. Compute (and plot) the 68% intervals with the following methods:
   1. central interval
   2. equal areas
   3. shortest interval
2. As a cross check, print the coverage of each method.
